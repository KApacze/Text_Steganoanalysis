Wyciągnięcie kodów znaków unicode i ich kategorii.  
Link do opisu kategorii: https://www.unicode.org/versions/Unicode15.0.0/ch04.pdf [strona 16]

```python
data_file = open(r"databases\UnicodeData.txt", 'r', encoding='utf-8')

unicode_characters = {}

for line in data_file:
    values = line.strip().split(';')
    hex_value = values[0]
    general_category = values[2]

    if general_category not in unicode_characters:
        unicode_characters[general_category] = set()
    unicode_characters[general_category].add(hex(int(hex_value, 16)))
data_file.close()                       
print(unicode_characters)            
```

```python
#Przeanalizuj wszystkie znaki z pliku i sprawdź stosunek każdej z kategorii
data_statistics = {"content": [],
                   "no_of_characters": [],
                    "no_of_printable_characters": [],
                   "no_of_words": [] }

for key, value in unicode_characters.items():
    data_statistics[f"${key}_ratio"] = []
```

```python

import unicodedata 
import string 
data_file = open("databases/test_review_dataset.csv", 'r')


next(data_file, None)
for line in data_file:
    no_of_characters = len(line)
    data_statistics["content"].append(line)
    data_statistics["no_of_characters"].append(no_of_characters)
    data_statistics["no_of_printable_characters"].append(sum(1 for char in line if char in string.printable))
    data_statistics["no_of_words"].append(len(line.split()))
    sum_per_category = {}
    for ch in line:
        category = unicodedata.category(ch)
        sum_per_category[category] = sum_per_category.get(category, 0) + 1
    for key, value in sum_per_category.items():
        data_statistics[f"${key}_ratio"].append([value/no_of_characters])


```
